```python
from bayes_opt import BayesianOptimization
```

```python
count = 0
```

```python
def objective(p1, p2, p3, p4, p5, p6): 
    global count  # Declare count as global   
    count = count + 1
    # return p1 + p2 + p3 + p4 + p5 + p6
    return (p1 * p2) / p3 + p4/p5 + p6 
```

```python
rf_bo = BayesianOptimization(
        objective,
        {'p1': (1, 10),
        'p2': (1, 10),
        'p3': (1, 100),
        'p4': (1, 100),
        'p5': (1, 100),
        'p6': (1, 100)
        }
    )
```

```python
rf_bo.maximize()
```

```python
index = []
for i in rf_bo.res:
    index.append(i['target'])
max_index = index.index(max(index))
print(rf_bo.res[max_index])
```
